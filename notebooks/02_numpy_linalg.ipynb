```python
# argmax, argsort, aggregation over axes
matrix = np.array([[1, 5, 3], [9, 2, 6]])
print(matrix.argmax())          # flattened index of max
print(matrix.argmax(axis=1))    # max index per row
print(matrix.argsort(axis=0))   # sorted indices per column
print(matrix.sum(axis=0))       # column-wise sum
print(matrix.mean(axis=1))      # row-wise mean
```

```python
# Mean-centre a matrix (subtract column mean from each column)
def mean_centre(X):
    return X - X.mean(axis=0)

X = np.random.rand(5, 3)
print(mean_centre(X))
```

```python
# Normalise rows to unit length
def normalise_rows(X):
    norms = np.linalg.norm(X, axis=1, keepdims=True)
    return X / norms

print(normalise_rows(X))
```

```python
# Pairwise Euclidean distances between two sets of points
def pairwise_distances(A, B):
    # A: (n, d), B: (m, d) -> output (n, m)
    diff = A[:, np.newaxis, :] - B[np.newaxis, :, :]
    return np.sqrt((diff ** 2).sum(axis=2))

A = np.random.rand(4, 2)
B = np.random.rand(3, 2)
print(pairwise_distances(A, B))
```

```python
# One-hot encode an integer label array
def one_hot(labels, num_classes):
    return np.eye(num_classes)[labels]

labels = np.array([0, 2, 1, 3])
print(one_hot(labels, 4))
```

```python
def cosine_similarity(X):
    # X: (n, d) -> returns (n, n) similarity matrix
    norm_X = normalise_rows(X)
    return norm_X @ norm_X.T

np.random.seed(42)
data_100x20 = np.random.rand(100, 20)
sim_matrix = cosine_similarity(data_100x20)

# Find two most similar rows (excluding self-similarity on diagonal)
np.fill_diagonal(sim_matrix, -1)  # ignore self-matches
i, j = np.unravel_index(sim_matrix.argmax(), sim_matrix.shape)
print(f"Most similar rows: {i} and {j}, similarity = {sim_matrix[i, j]:.4f}")
```

```python
import time

big_array = np.random.rand(1_000_000)

# Loop version
start = time.time()
result_loop = np.empty_like(big_array)
for k in range(len(big_array)):
    result_loop[k] = big_array[k] ** 2
loop_time = time.time() - start

# Vectorised version
start = time.time()
result_vec = big_array ** 2
vec_time = time.time() - start

print(f"Loop time: {loop_time:.4f}s")
print(f"Vectorised time: {vec_time:.6f}s")
print(f"Speedup: {loop_time / vec_time:.1f}x")
```

```python
A = np.random.rand(32, 128)
B = np.random.rand(128, 10)
b = np.random.rand(128)

# Predict shapes first (mentally), then verify:
product = A @ B
print("A @ B shape:", product.shape)   # expected (32, 10)

sum_result = A + b
print("A + b shape:", sum_result.shape)  # expected (32, 128), broadcasting
```

## Shape Drill Predictions

Given A (32, 128) and B (128, 10):

- **A @ B**: My prediction is (32, 10) — because matrix 
  multiplication removes the inner matching dimension (128) 
  and keeps the outer dimensions.

- **A + b** where b is (128,): My prediction is (32, 128) — 
  because NumPy broadcasts the 1D array `b` across every row 
  of A, since b's shape matches A's last dimension.

Verified in code below — both predictions were correct.

```python
def matmul_loops(A, B):
    n, m = A.shape
    m2, p = B.shape
    result = np.zeros((n, p))
    for i in range(n):
        for j in range(p):
            for k in range(m):
                result[i, j] += A[i, k] * B[k, j]
    return result

small_A = np.random.rand(20, 20)
small_B = np.random.rand(20, 20)

start = time.time()
r1 = matmul_loops(small_A, small_B)
t1 = time.time() - start

start = time.time()
r2 = small_A @ small_B
t2 = time.time() - start

start = time.time()
r3 = np.einsum('ik,kj->ij', small_A, small_B)
t3 = time.time() - start

print(f"Loops: {t1:.4f}s, @: {t2:.6f}s, einsum: {t3:.6f}s")
```
